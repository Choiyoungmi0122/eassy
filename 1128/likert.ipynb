```python
import matplotlib.pyplot as plt

import numpy as np
import seaborn as sns

# Example numeric data extracted from the responses
# Columns represent questions (e.g., Q7 to Q11) and rows represent participants
data = np.array([
    [2, 1, 2, 1, 6],  # Participant 1
    [5, 5, 5, 5, 5],  # Participant 2
    [1, 1, 1, 4, 6],  # Participant 3
    [2, 5, 3, 3, 2],  # Participant 4
])

# Question labels (Q7 to Q11)
questions = ["Q7", "Q8", "Q9", "Q10", "Q11"]

# Participant labels
participants = ["P1", "P2", "P3", "P4"]

# Create heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(data, annot=True, fmt="d", cmap="coolwarm", xticklabels=questions, yticklabels=participants, cbar_kws={'label': 'Score'})

# Title and labels
plt.title("Heatmap of Likert Scale Responses (Q7 to Q11)", fontsize=14)
plt.xlabel("Questions", fontsize=12)
plt.ylabel("Participants", fontsize=12)

# Show plot
plt.tight_layout()
plt.show()
```

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data preparation
data = {
    "Question": [f"Q{i}" for i in range(1, 15)],
    "A": [2, 1, 2, 1, 4, 2, 6, 6, 4, 2, 1, 1, 5, 1],
    "B": [5, 5, 5, 5, 5, 5, 5, 3, 5, 1, 3, 5, 6, 6],
    "C": [1, 1, 1, 4, 4, 4, 6, 1, 4, 1, 1, 2, 6, 1],
    "D": [2, 5, 3, 3, 2, 4, 2, 6, 2, 2, 4, 4, 6, 2],
}

# Create a DataFrame
df = pd.DataFrame(data)

# Melt the DataFrame for easier plotting with Seaborn
df_melted = df.melt(id_vars="Question", var_name="Category", value_name="Value")

# Plotting Likert scale interval plot
plt.figure(figsize=(12, 6))
sns.pointplot(data=df_melted, x="Question", y="Value", dodge=True, ci="sd", join=False, capsize=0.2)
plt.axhline(y=7, color='gray', linestyle='--')  # 7점 척도의 중간값
plt.axhline(y=6, color='gray', linestyle='--')
plt.axhline(y=5, color='gray', linestyle='--')
plt.axhline(y=4, color='gray', linestyle='--')
plt.axhline(y=3, color='gray', linestyle='--')
plt.axhline(y=2, color='gray', linestyle='--')
plt.axhline(y=1, color='gray', linestyle='--')
# Adding title and labels
# plt.title("Likert Scale Interval Plot")
# plt.xlabel("Questions")
# plt.ylabel("Mean Likert Score")
plt.xticks(rotation=0)
plt.tight_layout()

# Display the plot
plt.show()
```
