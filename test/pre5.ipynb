```python
import json
import pandas as pd 
import re

path = 'data/final_information.json'

with open(path, 'r', encoding="utf-8-sig") as f:
    data = json.load(f)

print(data[0:3])
# print(tpye(data))
```

```python
print(len(data))
```

```python
new_data = []

for item in data:
    temp = []
    for paragraph in item:
        # 숫자와 점으로 시작하는 부분을 찾아 새로운 리스트에 추가
        if paragraph.startswith('1.') or paragraph.startswith('2.') or paragraph.startswith('3.') or paragraph.startswith('4.'):
            temp.append(paragraph)
    new_data.append(temp)

print(new_data[0])
print(new_data[1])
print(new_data[2])
print(new_data)
```

```python
# 새로운 리스트 생성
new_list = []

# 각 섹션별로 나누어서 새로운 리스트에 추가
for section in data:
    new_section = []
    for item in section:
        # 각 섹션에서 번호를 기준으로 나눔
        splitted_item = item.split('\n\n')
        new_section.extend(splitted_item)
    new_list.append(new_section)

print(new_list[0][0])
```

```python
import re

# 새로운 리스트 생성
new_li = []

# 정규표현식 패턴
pattern = r'\d+\.'

# 각 섹션별로 나누어서 새로운 리스트에 추가
for section in data:
    new_section = []
    for item in section:
        # 정규표현식을 사용하여 데이터 분리
        splitted_item = re.split(pattern, item)
        # 빈 문자열을 제외하고 새로운 리스트에 추가
        new_section.extend([s.strip() for s in splitted_item if s.strip()])
    new_li.append(new_section)

print(new_li)
```
