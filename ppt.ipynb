```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# 데이터 값
percent = 65.5

# 도넛 차트 생성
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie([percent, 100-percent], startangle=90, colors=['#394F79', '#718DBE'], 
       wedgeprops={'linewidth': 10, 'edgecolor': 'white'})

# 가운데 텍스트 추가
plt.text(0, 0, f'{percent}%', ha='center', va='center', fontsize=40, color='#394F79')

# 가운데 빈 원 (도넛 모양을 만들기 위한 원)
centre_circle = Circle((0, 0), 0.70, color='white', fc='white', linewidth=0)
fig.gca().add_artist(centre_circle)

# 차트 스타일 제거
ax.axis('equal')

# 차트 보여주기
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for stacked bar chart
labels = ['합격자 자기소개서 참고', '무작정 많이 작성해봄', '주변의 참삭', '특별히 하는 것 없음', '자기소개서 관련 서적', '자기소개서 관련 강의 수강']
values = [50.3, 27.2, 23.2, 19.5, 10.3, 14.9]

# Stacked Bar Chart
fig, ax = plt.subplots()

bar_positions = np.arange(len(labels))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

# Horizontal bar chart
ax.barh(bar_positions, values, color=colors)

# Labels and title
ax.set_yticks(bar_positions)
ax.set_yticklabels(labels)
ax.set_xlabel('비율 (%)')
ax.set_title('자기소개서 작성의 어려움을 극복하기 위한 방안')

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import rc
rc('font', family='AppleGothic')
plt.rcParams['axes.unicode_minus'] = False
# Data for the chart
labels = ['자소서 작성 도구', 'GPT', '둘다 안씀']
sizes = [0, 75, 25]  # 0% for the first label as it's not represented in the chart
colors = ['#007bff', '#d9534f', '#f0ad4e']  # Custom colors for each section
explode = (0, 0.1, 0.1)  # 'explode' the 2nd and 3rd sections for emphasis

# Create the pie chart
plt.figure(figsize=(6,6))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.0f%%',
        shadow=True, startangle=140)

# Title
plt.title('자소서 생성시 보조 받을 수 있는 도구를 선택할때, 텍스트 생성기를 하나만 사용할 수 있다면 어떤 것을 사용하시겠습니까?')

# Show the chart
plt.show()
```
