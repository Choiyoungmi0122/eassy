```python
import json

# 파일을 불러오기
file_path = '/Users/choiyoungmi/Desktop/code/essay/data/information-allfinal.json'
with open(file_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 나눌 질문 항목 정의
question_types = [
    "지원동기", "경험", "성장과정", "성격의 장단점", "직무역량", "입사 후 포부", 
    "팀 활동 경험", "스트레스 관리 방법", "자신의 성격", "본인이 기여할 수 있는 부분", 
    "도전적 목표 설정 경험", "리더십 및 팔로우십 발휘 경험", "인재상에 부합하는 자신의 모습"
]

# 결과를 저장할 리스트
split_data = []

# 각 자기소개서를 질문과 답변으로 나누기
for essay in data:
    entry = {}
    entry['original'] = essay
    split_sections = {}
    
    # 각 질문 유형을 기준으로 분할
    for question in question_types:
        if question in essay[0]:
            split_sections[question] = essay[0].split(question, 1)[1].strip()

    entry['split'] = split_sections
    split_data.append(entry)

# 분할된 데이터를 json 파일로 저장
output_path = '/Users/choiyoungmi/Desktop/code/essay/data/information-allfinal_v2.json'
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(split_data, f, ensure_ascii=False, indent=4)

output_path
```

```python
import json
import re

# 원본 JSON 파일 불러오기
file_path = '/Users/choiyoungmi/Desktop/code/essay/data/information-allfinal.json'
with open(file_path, 'r', encoding='utf-8') as f:
    original_data = json.load(f)

# 새로운 데이터를 저장할 리스트
split_data = []



# 정규 표현식으로 찾을 패턴들
patterns = [
    r'.*십시오$', r'.*입니까\?$', r'.*서술하시오$', r'.*0자\)$', r'.*기술하여 주시기 바랍니다.$',
    r'.*소개해주세요$', r'.*설명해주십시오$', r'.*쓰시오$'
]

# \n\n\n 기준으로 분리하여 딕셔너리 형태로 저장
for entry in original_data:
    split_entry = entry[0].split('\n\n\n')
    split_dict = {}
    
    for section in split_entry:
        section = section.strip()
        key = "section"
        value = section
        
        for pattern in patterns:
            match = re.search(pattern, section)
            if match:
                key = match.group().strip()
                value = section[match.end():].strip()
                break
        
        split_dict[key] = value
    
    split_data.append(split_dict)



# 새로운 파일로 저장
output_split_path = '/Users/choiyoungmi/Desktop/code/essay/data/information-allfinal_V7.json'
with open(output_split_path, 'w', encoding='utf-8') as f:
    json.dump(split_data, f, ensure_ascii=False, indent=4)

output_split_path
```
