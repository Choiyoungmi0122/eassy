```python
import json

# Load the filtered JSON file
file_path = "/Users/choiyoungmi/Desktop/code/essay/data/information_web.json"


# Open and load the json file
with open(file_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

# Initialize a list to store the new format
formatted_data = []

# Iterate through the original data and transform into question-answer pairs
for entry in data:
    for key, value in entry.items():
        formatted_data.append({"question": key, "answer": value})

# Save the newly formatted data to a JSON file
new_file_path = '/Users/choiyoungmi/Desktop/code/essay/data/information_web_edit.json'
with open(new_file_path, 'w', encoding='utf-8') as f:
    json.dump(formatted_data, f, ensure_ascii=False, indent=4)

new_file_path  # Return the path to the saved formatted file
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the JSON file
df = pd.read_json('/Users/choiyoungmi/Desktop/code/essay/0910/0910_edit.json', encoding='utf-8')

# Save it as a CSV file
df.to_csv('/Users/choiyoungmi/Desktop/code/essay/0910/question_answer_dataset.csv', index=False, encoding='utf-8')

# Split the data into train, validation, and test sets
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)  # 70% train, 30% temp
valid_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)  # Split temp into 50% valid, 50% test

# Save the resulting datasets into separate CSV files
train_df.to_csv('/Users/choiyoungmi/Desktop/code/essay/0910/train_dataset_0910.csv', index=False, encoding='utf-8')
valid_df.to_csv('/Users/choiyoungmi/Desktop/code/essay/0910/valid_dataset_0910.csv', index=False, encoding='utf-8')
test_df.to_csv('/Users/choiyoungmi/Desktop/code/essay/0910/test_dataset_0910.csv', index=False, encoding='utf-8')
```
