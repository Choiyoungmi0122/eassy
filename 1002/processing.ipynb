```python
import re
import json
output_file_path = 'your_output_file_with_null_replaced.json'  # 저장할 파일 경로를 지정

with open(output_file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# 데이터에서 '내용'을 숫자. 기준으로 나누고, 빈 '내용' 필드는 삭제하는 코드
# 데이터에서 '내용'을 숫자. 기준으로 나누어 '내용1', '내용2' 등으로 저장하는 코드
for entry in data:
    if '내용' in entry and entry['내용'].strip():
        # '내용' 값을 "숫자." 기준으로 분할
        parts = re.split(r'(\d+\.\s)', entry['내용'])  # "숫자." 기준으로 분할
        
        # parts[0]이 1. 앞의 내용이면 무시하고, 1. 이후부터 시작
        if parts[0].strip():  # 1. 앞의 내용이 있으면 그대로 '내용'에 저장
            entry['내용1'] = parts[0].strip()
        else:
            entry['내용1'] = (parts[1] + parts[2]).strip()  # 1.부터 시작하는 내용

        # 나머지 분할된 내용을 '내용2', '내용3', ...으로 저장
        for i in range(3, len(parts), 2):
            section_number = i // 2 + 1  # '내용2', '내용3' ...를 위한 숫자 계산
            entry[f'내용{section_number}'] = (parts[i] + parts[i + 1]).strip()

        # 원본 '내용' 필드는 제거
        del entry['내용']


# 수정된 데이터를 파일로 저장
output_file_path = 'split_content_first.json'  # 저장할 파일 경로를 지정
with open(output_file_path, 'w', encoding='utf-8') as file:
    json.dump(data, file, indent=4, ensure_ascii=False)
```

```python
import json

# 첫 번째 파일 로드
input_file_path = '/Users/choiyoungmi/Desktop/code/essay/1002/personality_traits_v3.json'  # 입력 파일 경로
output_file_path = '/Users/choiyoungmi/Desktop/code/essay/1002/motivation_v1.json'  # 결과를 저장할 파일 경로

# 입력 파일 읽기
with open(input_file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# 데이터에서 특정 키를 '지원동기'로 변경하는 코드
for entry in data:
    keys_to_change = {}
    
    # 기존 키를 순회하며 '[입사 지원동기]' 또는 '입사 지원동기'를 찾아서 변경
    for key in list(entry.keys()):
        if key in ['[입사 지원동기]', '[지원동기]']:
            # 새로운 '지원동기' 키로 값을 저장
            keys_to_change['지원동기'] = entry[key]
            # 기존 키 삭제
            del entry[key]
    
    # 새롭게 '지원동기'로 추가
    entry.update(keys_to_change)

# 수정된 데이터를 파일로 저장
with open(output_file_path, 'w', encoding='utf-8') as file:
    json.dump(data, file, indent=4, ensure_ascii=False)
```

```python
import json
import os

# JSON 파일 로드
input_file_path = '/Users/choiyoungmi/Desktop/code/essay/1002/remind_3.json'  # 원본 파일 경로
output_dir = '/Users/choiyoungmi/Desktop/code/essay/1002/data/'  # 저장할 디렉터리 경로
base_output_filename = 'remind_'  # 파일 이름의 베이스

with open(input_file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# 200개씩 나누어서 저장
chunk_size = 200
total_chunks = (len(data) + chunk_size - 1) // chunk_size  # 총 몇 개의 파일이 생성될지 계산

for i in range(total_chunks):
    # 데이터 슬라이스
    start_index = i * chunk_size
    end_index = (i + 1) * chunk_size
    sliced_data = data[start_index:end_index]

    # 파일명 설정 (remind_4부터 시작하도록 설정)
    output_file_path = os.path.join(output_dir, f"{base_output_filename}{i + 4}.json")
    
    # 슬라이스된 데이터를 파일로 저장
    with open(output_file_path, 'w', encoding='utf-8') as file:
        json.dump(sliced_data, file, indent=4, ensure_ascii=False)

    print(f"데이터 {start_index}부터 {end_index}까지 항목이 '{output_file_path}'에 저장되었습니다.")
```
