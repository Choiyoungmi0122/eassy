```python
import json

# 필수 key 값 목록
required_keys = [
    "지원동기", "입사 후 포부", "성장 과정", "성격의 장단점", 
    "직무 역량", "사회 이슈", "극복 경험", "협업 경험", 
    "자유 양식", "기타"
]
path = "/Users/choiyoungmi/Desktop/code/essay/1002/Ryu_data.json"
# 각 entry에 대해 필수 key를 검사하고 누락된 경우 '없음'으로 채움
def ensure_required_keys(entry, required_keys):
    for key in required_keys:
        if key not in entry:
            entry[key] = "없음"
    return entry

# JSON 파일에서 데이터를 불러옴 (combined_data로 불러오기)
input_file_path = path
with open(input_file_path, "r", encoding="utf-8") as f:
    combined_data = json.load(f)

# 필수 key가 포함되도록 cleaned_data 생성
cleaned_data = [ensure_required_keys(entry, required_keys) for entry in combined_data]

# cleaned_data를 저장할 경로
final_output_file_retry = path

# cleaned_data 저장
with open(final_output_file_retry, "w", encoding="utf-8") as f:
    json.dump(cleaned_data, f, ensure_ascii=False, indent=4)

# sliced_data에 대해 동일한 처리
sliced_data = cleaned_data[:100]  # 예시로 일부 데이터를 slicing

# sliced_data를 저장할 경로
saved_path = path

# sliced_data 저장
with open(saved_path, 'w', encoding='utf-8') as file:
    json.dump(sliced_data, file, indent=4, ensure_ascii=False)
```
